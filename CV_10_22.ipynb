```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow
from PIL import Image

# Upload image
uploaded = files.upload()

# Get uploaded filename
filename = list(uploaded.keys())[0]

# Read image using PIL
img = Image.open(filename)

# Convert PIL image to OpenCV format
img = np.array(img)
img = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)

# Get image size
rows, cols = img.shape[:2]

# Translation matrix (Affine Transformation)
M = np.float32([[1, 0, 80],
                [0, 1, 40]])

# Apply affine transformation
affine_img = cv2.warpAffine(img, M, (cols, rows))

# Display images
print("Original Image")
cv2_imshow(img)

print("Affine Transformed Image")
cv2_imshow(affine_img)

# Save output
cv2.imwrite("Affine_Transformed.jpg", affine_img)
```

```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read image
img = cv2.imread(filename)

h, w = img.shape[:2]

# Source points
pts1 = np.float32([
    [0, 0],
    [w, 0],
    [0, h],
    [w, h]
])

# Destination points
pts2 = np.float32([
    [70, 0],          # Top-left
    [w-70, 0],        # Top-right
    [90, h],          # Bottom-left
    [w-90, h]         # Bottom-right
])

# Perspective transform
matrix = cv2.getPerspectiveTransform(pts1, pts2)

result = cv2.warpPerspective(img, matrix, (w, h))

print("Original Image")
cv2_imshow(img)

print("Perspective Transformed Image")
cv2_imshow(result)

cv2.imwrite("Perspective_Output.jpg", result)
```

```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Read image
filename = list(uploaded.keys())[0]
img = cv2.imread(filename)

# Get image dimensions
h, w = img.shape[:2]

# Source points
pts1 = np.float32([
    [0, 0],
    [w-1, 0],
    [0, h-1],
    [w-1, h-1]
])

# Destination points
pts2 = np.float32([
    [60, 0],
    [w-60, 0],
    [20, h-1],
    [w-20, h-1]
])

# Perspective transformation matrix
matrix = cv2.getPerspectiveTransform(pts1, pts2)

# Apply perspective transformation
result = cv2.warpPerspective(img, matrix, (w, h))

# Display images
print("Original Image")
cv2_imshow(img)

print("Perspective Transformed Image")
cv2_imshow(result)

# Save the result
cv2.imwrite("Perspective_Output.jpg", result)
```

```python
# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Gaussian Blur
blur = cv2.GaussianBlur(gray, (5,5), 0)

# Increase contrast
blur = cv2.equalizeHist(blur)

# Canny Edge Detection
edges = cv2.Canny(blur, 20, 80)

# Slightly thicken the edges
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (2,2))
edges = cv2.dilate(edges, kernel, iterations=1)

print("Output")
cv2_imshow(edges)

cv2.imwrite("Edge_Detected.jpg", edges)
```
