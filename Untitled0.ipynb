```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get uploaded file name
filename = list(uploaded.keys())[0]

# Read image
img = cv2.imread(filename)

# Convert image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Canny Edge Detection
edges = cv2.Canny(gray, 100, 200)

# Display edge image
cv2_imshow(edges)

# Save output image
cv2.imwrite("Canny_Edges.jpg", edges)

# Download output image
files.download("Canny_Edges.jpg")
```

```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get uploaded file name
filename = list(uploaded.keys())[0]

# Read image in grayscale
img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

# Create a 5x5 kernel
kernel = np.ones((5, 5), np.uint8)

# Dilate the image
dilated_img = cv2.dilate(img, kernel, iterations=1)

# Display the dilated image
cv2_imshow(dilated_img)

# Save the output image
cv2.imwrite("Dilated_Image.jpg", dilated_img)

# Download the output image
files.download("Dilated_Image.jpg")
```

```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get uploaded file name
filename = list(uploaded.keys())[0]

# Read image
img = cv2.imread(filename)

# Convert image to grayscale
gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Create a 5x5 rectangular kernel
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

# Apply erosion
eroded_image = cv2.erode(gray_image, kernel, iterations=1)

# Display the eroded image
cv2_imshow(eroded_image)

# Save the output image
cv2.imwrite("eroded_image.jpg", eroded_image)

# Download the output image
files.download("eroded_image.jpg")
```

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get uploaded file name
filename = list(uploaded.keys())[0]

# Read image
image = cv2.imread(filename)

# Get image dimensions
height, width = image.shape[:2]

# Resize to bigger image (3x)
scale_factor = 3.0
bigger_image = cv2.resize(image, (int(width * scale_factor), int(height * scale_factor)))

# Resize to smaller image (0.5x)
scale_factor = 0.5
smaller_image = cv2.resize(image, (int(width * scale_factor), int(height * scale_factor)))

# Display images
print("Original Image")
cv2_imshow(image)

print("Bigger Image")
cv2_imshow(bigger_image)

print("Smaller Image")
cv2_imshow(smaller_image)

# Save output images
cv2.imwrite("Bigger_image.jpg", bigger_image)
cv2.imwrite("Smaller_image.jpg", smaller_image)

# Download output images
files.download("Bigger_image.jpg")
files.download("Smaller_image.jpg")
```

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get uploaded file name
filename = list(uploaded.keys())[0]

# Read image
img = cv2.imread(filename)

# Rotate clockwise
clockwise_img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

# Rotate counterclockwise
counterclockwise_img = cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE)

# Display images
print("Original Image")
cv2_imshow(img)

print("Clockwise Rotated Image")
cv2_imshow(clockwise_img)

print("Counterclockwise Rotated Image")
cv2_imshow(counterclockwise_img)

# Save images
cv2.imwrite("clockwise_image.jpg", clockwise_img)
cv2.imwrite("counterclockwise_image.jpg", counterclockwise_img)

# Download images
files.download("clockwise_image.jpg")
files.download("counterclockwise_image.jpg")
```
